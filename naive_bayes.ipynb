```python
import numpy as np
import pandas as pd
df = pd.read_csv('https://gist.githubusercontent.com/DiogoRibeiro7/c6590d0cf119e87c39e31c21a9c0f3a8/raw/4a8e3da267a0c1f0d650901d8295a5153bde8b21/PlayTennis.csv')
df.head()
print(pd.crosstab(df.Outlook,df['Play Tennis'],normalize='columns'))
print('          ')
print(pd.crosstab(df.Temperature,df['Play Tennis'],normalize='columns'))
print('          ')
print(pd.crosstab(df.Humidity,df['Play Tennis'],normalize='columns'))
print('          ')
print(pd.crosstab(df.Wind,df['Play Tennis'],normalize='columns'))
```
