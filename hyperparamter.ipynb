```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv')
df.head()
X=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
df.info()
```

```python
df.describe()
```

```python
df.duplicated().sum()
```

```python
from sklearn.model_selection import cross_val_score,KFold
from sklearn.neighbors import KNeighborsRegressor
```

```python
knn=KNeighborsRegressor()
kfold = KFold(n_splits=5, shuffle=True, random_state=1)
scores = cross_val_score(knn, X, y, cv=kfold, scoring='r2')
scores.mean()
```

```python
param_grid={'weights':['uniform','distance'],
            'algorithm':['ball_tree', 'kd_tree', 'brute'],
            'p':[1,2],
            'n_neighbors':[1,3,5,7,9,11]
            }
```

```python
from sklearn.model_selection import GridSearchCV
gc=GridSearchCV(estimator=knn,param_grid=param_grid,scoring='r2', n_jobs=-1, refit=True, cv=kfold, verbose=0, pre_dispatch='2*n_jobs',  return_train_score=False)
```

```python
gc.fit(X,y)
```

```python
print(gc.best_params_)
print(gc.best_score_)
```

```python
pd.DataFrame(gc.cv_results_)[['param_algorithm',	'param_n_neighbors',	'param_p', 'param_weights', 'mean_test_score']].sort_values('mean_test_score',ascending=False).head(10)
```

```python
from sklearn.model_selection import RandomizedSearchCV
rv=RandomizedSearchCV(estimator=knn,param_distributions=param_grid,scoring='r2', n_jobs=-1, refit=True, cv=kfold, verbose=0, pre_dispatch='2*n_jobs',  return_train_score=False)
```

```python
rv.fit(X,y)
```

```python
rv.best_params_
```
