```python
import numpy as np

from sklearn.tree import DecisionTreeRegressor


class GradientBoostingClassifierScratch:

    def __init__(
        self,
        n_estimators=5,
        learning_rate=0.1,
        max_depth=2
    ):

        self.n_estimators = n_estimators

        self.learning_rate = learning_rate

        self.max_depth = max_depth

        self.trees = []

        self.leaf_values = []

        self.initial_prediction = None


    # ======================================
    # SIGMOID
    # ======================================

    def sigmoid(self, x):

        return 1 / (
            1 + np.exp(-x)
        )


    # ======================================
    # FIT
    # ======================================

    def fit(self, X, y):

        # ----------------------------------
        # INITIAL PROBABILITY
        # ----------------------------------

        p = np.mean(y)


        p = np.clip(
            p,
            1e-15,
            1 - 1e-15
        )


        # ----------------------------------
        # INITIAL LOG-ODDS
        # ----------------------------------

        self.initial_prediction = np.log(

            p / (1 - p)

        )


        # Initial raw predictions

        F = np.full(

            len(y),

            self.initial_prediction,

            dtype=float

        )


        # ==================================
        # BOOSTING ITERATIONS
        # ==================================

        for m in range(

            self.n_estimators

        ):


            print("\n" + "=" * 60)

            print(

                f"TREE {m + 1}"

            )

            print("=" * 60)


            # ----------------------------------
            # CURRENT PROBABILITIES
            # ----------------------------------

            p = self.sigmoid(F)


            # ----------------------------------
            # PSEUDO-RESIDUALS
            # ----------------------------------

            residuals = y - p


            print(

                "\nCurrent probabilities:"

            )

            print(

                np.round(p, 3)

            )


            print(

                "\nPseudo-residuals:"

            )

            print(

                np.round(residuals, 3)

            )


            # ----------------------------------
            # TRAIN TREE ON RESIDUALS
            # ----------------------------------

            tree = DecisionTreeRegressor(

                max_depth=self.max_depth,

                random_state=42

            )


            tree.fit(

                X,

                residuals

            )


            # ----------------------------------
            # FIND LEAF FOR EVERY SAMPLE
            # ----------------------------------

            leaf_ids = tree.apply(X)


            # Store optimal values
            leaf_value_dict = {}


            # ----------------------------------
            # CALCULATE GAMMA FOR EACH LEAF
            # ----------------------------------

            for leaf in np.unique(

                leaf_ids

            ):


                # Samples belonging to leaf

                indices = (

                    leaf_ids == leaf

                )


                # Numerator

                numerator = np.sum(

                    y[indices]

                    -

                    p[indices]

                )


                # Denominator

                denominator = np.sum(

                    p[indices]

                    *

                    (

                        1

                        -

                        p[indices]

                    )

                )


                # Avoid division by zero

                if denominator == 0:

                    gamma = 0


                else:

                    gamma = (

                        numerator

                        /

                        denominator

                    )


                leaf_value_dict[leaf] = gamma


            # ----------------------------------
            # CREATE CORRECTIONS
            # ----------------------------------

            corrections = np.array([

                leaf_value_dict[leaf]

                for leaf in leaf_ids

            ])


            print(

                "\nLeaf corrections:"

            )

            print(

                np.round(corrections, 3)

            )


            # ----------------------------------
            # UPDATE MODEL
            # ----------------------------------

            F = (

                F

                +

                self.learning_rate

                *

                corrections

            )


            print(

                "\nUpdated log-odds:"

            )

            print(

                np.round(F, 3)

            )


            # ----------------------------------
            # STORE MODEL
            # ----------------------------------

            self.trees.append(

                tree

            )


            self.leaf_values.append(

                leaf_value_dict

            )


        return self


    # ======================================
    # PREDICT PROBABILITY
    # ======================================

    def predict_proba(self, X):


        # Initial log-odds

        F = np.full(

            X.shape[0],

            self.initial_prediction,

            dtype=float

        )


        # Add every tree

        for tree, leaf_dict in zip(

            self.trees,

            self.leaf_values

        ):


            # Find leaves

            leaf_ids = tree.apply(X)


            # Get correction for leaf

            corrections = np.array([

                leaf_dict[leaf]

                for leaf in leaf_ids

            ])


            # Update

            F += (

                self.learning_rate

                *

                corrections

            )


        # Convert log-odds → probability

        probabilities = self.sigmoid(F)


        return probabilities


    # ======================================
    # PREDICT CLASS
    # ======================================

    def predict(self, X):


        probabilities = (

            self.predict_proba(X)

        )


        return (

            probabilities >= 0.5

        ).astype(int)
```
